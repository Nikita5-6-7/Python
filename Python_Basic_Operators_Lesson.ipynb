# ➕ Python Basic Operators

Operators tell Python to **perform an action on values**.

Example: `10 + 5` — here, `+` is the operator.

## 🧮 1. Arithmetic Operators

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 5` |
| `-` | Subtraction | `10 - 5` |
| `*` | Multiplication | `10 * 5` |
| `/` | Division | `10 / 5` |
| `//` | Floor division | `10 // 3` |
| `%` | Remainder | `10 % 3` |
| `**` | Power | `2 ** 3` |

```python
a = 10
b = 3

print('Addition:', a + b)
print('Subtraction:', a - b)
print('Multiplication:', a * b)
print('Division:', a / b)
print('Floor division:', a // b)
print('Remainder:', a % b)
print('Power:', a ** b)
```

### 💡 The `%` operator

`%` gives the remainder after division. It can also help check whether a number is even.

```python
num1 = 20
num2 = 3
print("Addition" , num1 + num2)
print("Subtraction", num1-num2)
print("Reminder", num1%num2)
print ("Division", num1/num2)
print("floor division", num1//num2)
print("multiplication", num1 *num2)
print("Power" , num1 ** num2)
```

```python
number = 8
print(number % 2)
```

## ⚖️ 2. Comparison Operators

Comparison operators return `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

```python
x = 10
y = 5

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

### ⚠️ Important

`=` assigns a value: `score = 10`

`==` compares values: `score == 10`

## 🔄 3. Assignment Operators

Assignment operators store or update values.

```python
score = 10

score += 5
print(score)
score -= 2
print(score)
score *= 2
print(score)
```

```python
marks = 20

marks += 20
print(marks)
marks -= 20
print(marks)
```

## 🧠 4. Logical Operators

The main logical operators are `and`, `or`, and `not`.

```python
age = 20
print(age >= 18 and age <= 30)
print(age > 20 and age < 30)
print(age > 20 or age < 30)

day = 'Saturday'
print(day == 'Saturday' or day == 'Sunday')

is_raining = False
print(not is_raining)
```

## 🌍 Real-World Example

```python
price = 20
quantity = 3

total = price * quantity
is_expensive = total > 50

print('Total:', total)
print('Is expensive:', is_expensive)
```
