# 🚦 Python Conditionals

Conditionals let Python **make decisions** by checking whether something is `True` or `False`.

## 🧠 1. `if` Statement

Use `if` when code should run only when a condition is true.

```python
age = 20

if age >= 22:
    print("You are an adult")
else:
    print("you are not adult")
```

### ⚠️ Indentation Matters

Code inside an `if` statement must be indented.

```python
temperature = 30

if temperature > 25:
    print("It is hot today")
elif temperature >=30:
    print("It is very hop")
else:
  print("It is Normal Temperature")
```

## 🔀 2. `if` and `else`

Use `else` when you want another action if the condition is false.

```python
age = 16

if age >= 18:
    print("You are an adult")
else:
    print("You are under 18")
```

## 🪜 3. `elif`

Use `elif` when there are more than two possible choices.

```python
score = 75

if score >= 80:
    print("Excellent")
elif score >= 50:
    print("Pass")
else:
    print("Try again")
```
