# 📚 Python Lists and Dictionaries

Lists and dictionaries help us store **multiple pieces of information** in one place.


## ❓ Why Do We Need Them?

Imagine storing three student names using separate variables:

```python
student1 = "Maria"
student2 = "Ahmed"
student3 = "Sara"
```

This becomes difficult when we have many values.

Lists and dictionaries let us **group related data together**, making code easier to manage and reuse.

# 🧠 1. Lists

A **list** stores multiple values in one variable.

Lists use square brackets: `[]`

````python
students = ["Maria", "Ahmed", "Sara"]

print(students)
````

## 🔎 Accessing List Items

Each item has a position called an **index**.

Python starts counting from `0`.

````python
students = ["Maria", "Ahmed", "Sara"]

print(students[0])
print(students[1])
````

````python
city = ["Toronto", "Mississauga","Alberta", "Calgary"]
print(city[1:3])
print(city[0:4])
print(city[2:5])
print(city[1:4])
````

## ➕ Adding an Item

Use `.append()` to add a new item.

Append use to add new ites at the end of the list.

````python
students = ["Maria", "Ahmed"]

students.append("Sara")
students.append("Nikita")

print(students)
````

insert() method is user to add items at specific place.


````python
fruits = ["Apple", "Orange", "Grapes", "Banana"]
fruits.insert(2,"jackfruits")
print(fruits)
````

To append elements from another list to the current list, use the extend() method.

````python
code = [102, 103, 104]
name = ["nikita", "krishna", "hardip"]
code.extend(name)
print(code)
````

## ✏️ Changing an Item

````python
students = ["Maria", "Ahmed", "Sara"]

students[1] = "John"
students[0:2]=["Nikita", "Shalini"]

print(students)
````

## 🌍 Why Are Lists Useful?

Lists are useful when you have **many similar values**.

Examples:

- student names
- shopping cart items
- test scores
- cities
- temperatures

# 🗂️ 2. Dictionaries

A **dictionary** stores information using **key-value pairs**.

Think of it like a real dictionary:

```text
name  → Maria
age   → 21
city  → Toronto
```

Dictionaries use curly brackets: `{}`

````python
student = {
    "name": "Maria",
    "age": 21,
    "city": "Toronto"
}

print(student)
````

## 🔎 Accessing Dictionary Values

Use the key to access its value.

You can access the items of a dictionary by referring to its key name, inside square brackets.
There is also a method called **get()** that will give you the same result.
The **keys()** method will return a list of all the keys in the dictionary.
The **values()** method will return a list of all the values in the dictionary.
The **items()** method will return each item in a dictionary, as tuples in a list.
To determine if a specified key is present in a dictionary use the in keyword.

````python
student = [
  {
      "name": "Maria",
    "age": 21,
    "city": "Toronto"
  },
  {
  "name" : "Alex",
  "age" : 22,
  "city" : "Toronto"
  },
  {
      "name" : "Sara",
      "age"  : 21,
      "city" : "Mississauga"
  }
]
print(student)
````

````python
student = {
    "name": "Maria",
    "age": 21
}

print(student["name"])
print(student["age"])
x = student.get("name")
print(x)
````

````python
student = [
{
      "name": "Maria",
    "age": 21,
    "city": "Toronto"
  },
  {
  "name" : "Alex",
  "age" : 22,
  "city" : "Toronto"
  },
  {
      "name" : "Sara",
      "age"  : 21,
      "city" : "Mississauga"
  }
]
print(student[0]["name"])
print(student[1]["age"])
````

## ➕ Adding New Information

````python
student = {
    "name": "Maria",
    "age": 21
}

student["course"] = "Python"
student["city"] = "Toronto"
student["marks"] = 80

print(student)
````

## ✏️ Updating a Value

The update() method will update the dictionary with the items from a given argument. If the item does not exist, the item will be added.

The argument must be a dictionary, or an iterable object with key:value pairs.

````python
student = {
    "name": "Maria",
    "age": 21
}

student["age"] = 22

print(student)
````

````python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict.update({"model": "BMW"})
thisdict.update({"color" : "Red"})
print(thisdict)
````

````python
myfamily = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007
  },
  "child3" : {
    "name" : "Linus",
    "year" : 2011
  }
}
x = myfamily["child1"]["name"]
print(x)

for x, obj in myfamily.items():
  print(x)
  for y in obj:
    print(y + ":", obj[y])
````

## 🌍 Why Are Dictionaries Useful?

Dictionaries are useful when values need **labels**.

Examples:

- user profile
- product information
- student record
- contact details
- settings

# ⚖️ List vs Dictionary

| List | Dictionary |
|---|---|
| Stores a collection of values | Stores key-value pairs |
| Uses `[]` | Uses `{}` |
| Access with an index | Access with a key |
| Good for similar items | Good for labelled information |


## 🌍 Real-World Example

````python
student = {
    "name": "Maria",
    "scores": [80, 90, 85]
}

print(student["name"])
print(student["scores"])
````

Here, the dictionary stores information about one student, while the list stores multiple scores.
