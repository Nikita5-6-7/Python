# Variables

A variables is a name that refers to a value in python.


Think of it like a labelled box. The label is the variable name, and the value is what you've placed inside.



*   Label : student_name
*   Contents: Maria

In Python,







```python
student_name = "Maria"
student_name
```

Here,

*   student_name  → variable name
*   =             → assignment operator


*   "Maria"       → value


So essentially, python is like:


“Store "Maria" and let me refer to it using the name student_name.”



# **WHY Do Variables Exist?**

Imagine you're writing a program for a student called Maria.

Without a variable, you might write:

```python
print("Hello, Maria!")
print("Welcome to Python, Maria!")
print("Maria, your lesson is ready.")
```

This works, but what happens when a different student uses the program?

You would have to find "Maria" everywhere and replace it.

Instead, we can create a variable:

```python
student_name = "Maria"

print("Hello,", student_name)
print("Welcome to Python,", student_name)
print(student_name, "your lesson is ready.")
```

```python
# Now change the student_ name to your name and run it again

student_name = ""

print("Hello,", student_name)
print("Welcome to Python,", student_name)
print(student_name, "your lesson is ready.")
```

**Variables make programs:**

Reusable — store a value once and use it many times.

Flexible — change the variable and the rest of the program can use the new value.

Readable — student_name tells us what the value represents.

Easier to maintain — we don't need to repeat the same information throughout our code.

# **Variables Can Change**

```python
score = 10

print(score)
```

The variable still has the same name **score** but the value has changed from 10 -> 20.

# **= Does NOT Mean “Equals” Here**

**In mathematics:**

x = 10

usually means:

x is equal to 10.



**In Python assignment:**

x = 10

means:

“Assign the value 10 to the variable x.”





Later, when we learn comparison operators, we'll see that Python uses:

==

to ask whether two values are equal.


# **Naming Variables**

Variable names should clearly describe what they contain.

✅  Good names

student_name = "Maria"

student_age = 21

total_score = 95

course_name = "Python"


❌ Poor names

x = "Maria"

a = 21

thing = 95

abc = "Python"


The second group technically works, but another programmer has no idea what those variables represent.

Compare:

x = 95

with:

exam_score = 95

Which one is easier to understand?

# **Python Is Case-Sensitive**

Python treats uppercase and lowercase letters as different.

name = "Maria"

Name = "Ahmed"


print(name)

print(Name)

Output:

Maria

Ahmed

These are two different variables:

name ≠ Name

```python
name = "Maria"
Name = "Ahmed"

print(name)
print(Name)
```
