# 🧠 Python Data Types

A **data type** tells Python what kind of value a variable contains.

For example:

```python
name = "Maria"      # str or string
age = 25            # int or integer
height = 1.68       # float or decimal
is_student = True   # bool or boolean

print(name, age, height, is_student)
```

Python needs to know the type because different kinds of data behave differently.

---

## ❓ WHY Do Data Types Matter?

Compare these two examples:

```python
print(10 + 5)
```

The result is:

`15`

Now compare it with strings:

```python
print("10" + "5")
```

The result is:

`105`

Why?

- `10` and `5` are **numbers**, so Python adds them.
- `"10"` and `"5"` are **strings**, so Python joins them.

The data type determines **what Python can do with a value**.

---

## 🔍 Essential Python Data Types

For now, focus on these four:

| Type | Meaning | Example |
|---|---|---|
| `str` | Text | `"Hello"` |
| `int` | Whole number | `25` |
| `float` | Decimal number | `19.99` |
| `bool` | True or False | `True` |

### 📝 `str` — String

A **string** stores text and uses quotation marks.

```python
name = "Maria"
city = "Toronto"

print(name)
print(city)
```

Even numbers become strings when placed inside quotes:

```python
phone_code = "416"

print(phone_code)
print(type(phone_code))
```

`"416"` is text, not a number.

### 🔢 `int` — Integer

An **integer** is a whole number without a decimal point.

```python
age = 25
score = 90
items = 5
temperature = -10

print(age, score, items, temperature)
```

### 🔢 `float` — Decimal Number

A **float** is a number containing a decimal point.

```python
price = 19.99
height = 1.68
temperature = 24.5

print(price, height, temperature)
```

### 🔘 `bool` — Boolean

A **Boolean** has only two possible values:

`True` or `False`

Booleans are useful for yes/no or true/false decisions.

```python
is_student = True
is_raining = False

print(is_student)
print(is_raining)
```

---

## 🔎 Checking a Data Type with `type()`

Python's `type()` function tells us the data type of a value.

```python
name = "Maria"
age = 25
height = 1.68
is_student = True

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

---

## 🌍 Real-World Usage

```python
username = "maria123"    # str
items_in_cart = 3        # int
cart_total = 49.99       # float
is_logged_in = True      # bool

print(username)
print(items_in_cart)
print(cart_total)
print(is_logged_in)
```

Choosing the correct data type helps Python understand how the information should be used.
